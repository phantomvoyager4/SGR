```python
import pandas as pd
import umap
import plotly.express as px
import time
from sklearn.cluster import DBSCAN
from sklearn.decomposition import TruncatedSVD
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

## Load data

`load_data` reads the pre-vectorized game dataset from a parquet file into a DataFrame; everything downstream (embedding, clustering, plotting) works off this one table.

- **path** — filesystem path to the parquet file; defaults to the vectorized games dataset checked into `data/parquet/`.
- **df** — the loaded DataFrame, one row per game and one column per feature/metadata field, returned unchanged.

The cell below also splits `df` into `metadata_table` (identifier columns like `name`, kept only for display/hover) and `vectors_table` (everything else — the feature columns that get embedded).

```python
def load_data(path='../../../data/parquet/games_fully_vectorized.parquet'):
    df = pd.read_parquet(path)
    return df

df = load_data()


potential_metadata = ['index', 'name']
metadata_cols = [col for col in potential_metadata if col in df.columns]

metadata_table = df[metadata_cols].copy()

vectors_table = df.drop(columns=metadata_cols)
```

## Build the 2D embedding

`build_embedding` reduces the high-dimensional feature vectors to 2D for plotting. It runs a fast linear TruncatedSVD pass first to cut dimensionality and denoise, then hands that off to UMAP to fold it into 2D while preserving local neighborhood structure. Both stages share a `random_state`, so re-running this cell reproduces the same layout instead of a different one every time.

- **numerical_vectors** — the numeric feature matrix (one row per game) to embed; non-numeric columns (including booleans — pandas doesn't count `bool` as `number`) were already dropped by `select_dtypes`, so it's worth checking the dropped-column count isn't hiding a feature you actually wanted.
- **n_svd_components** — how many components TruncatedSVD keeps before handing off to UMAP; keeps UMAP's input compact and less noisy.
- **random_state** — seed shared by SVD and UMAP so reruns reproduce the same embedding. Note: setting this makes UMAP ignore multi-core parallelism (it forces single-threaded execution once seeded), which is why `n_jobs` isn't passed here anymore.
- **svd** — the fitted TruncatedSVD reducer.
- **vectors_reduced** — `numerical_vectors` projected onto `n_svd_components` dimensions.
- **reducer** — the fitted UMAP reducer.
- **embedding** — the final `(UMAP_X, UMAP_Y)` array returned to the caller and stored in `plot_df`.

```python
def build_embedding(numerical_vectors, n_svd_components=100, random_state=42):
    print(f"[{time.strftime('%X')}] Step 1/3: Reducing dimensions via TruncatedSVD...")
    svd = TruncatedSVD(n_components=n_svd_components, random_state=random_state)
    vectors_reduced = svd.fit_transform(numerical_vectors)

    print(f"[{time.strftime('%X')}] Step 2/3: Initializing UMAP...")
    reducer = umap.UMAP(
        n_neighbors=15,
        min_dist=0.1,
        metric='cosine',
        init='spectral',
        random_state=random_state,
    )

    print(f"[{time.strftime('%X')}] Step 3/3: Fitting UMAP over {len(numerical_vectors):,} rows...")
    embedding = reducer.fit_transform(vectors_reduced)

    print(f"[{time.strftime('%X')}] Processing complete. Ready for visualization.")
    return embedding


numerical_vectors = vectors_table.select_dtypes(include=['number'])

missing_cols = len(vectors_table.columns) - len(numerical_vectors.columns)
if missing_cols > 0:
    print(f"Removed {missing_cols} non-numeric/sequence columns from feature space.")

umap_embedding = build_embedding(numerical_vectors)

plot_df = metadata_table.copy()
plot_df['UMAP_X'] = umap_embedding[:, 0]
plot_df['UMAP_Y'] = umap_embedding[:, 1]
```

## Plot the embedding

`plot_umap` renders a 2D embedding as a WebGL scatter plot with consistent styling (dark theme, small markers), so both the plain projection below and the cluster-colored plot further down share one definition instead of drifting apart.

- **df** — the plotting DataFrame; must contain `UMAP_X`/`UMAP_Y` columns.
- **color** — optional column name to color points by (e.g. a cluster id column); `None` renders every point the same color.
- **color_discrete_map** — optional explicit `{value: color}` mapping, used later for the cluster plot so specific cluster ids and noise get controlled colors.
- **title** — plot title.
- **marker_size** / **opacity** — marker styling; the cluster plot below overrides `opacity` to `1.0` so distinct cluster colors stay easy to tell apart.
- **hover_data** — extra columns shown on hover; defaults to the non-name metadata columns.
- **fig** — the constructed Plotly figure, shown and returned.

```python
def plot_umap(df, color=None, color_discrete_map=None, title='', marker_size=2, opacity=0.5, hover_data=None):
    fig = px.scatter(
        df,
        x='UMAP_X',
        y='UMAP_Y',
        color=color,
        hover_name='name' if 'name' in df.columns else None,
        hover_data=hover_data if hover_data is not None else [col for col in metadata_cols if col != 'name'],
        color_discrete_map=color_discrete_map,
        title=title,
        render_mode='webgl',
    )
    fig.update_traces(marker=dict(size=marker_size, opacity=opacity))
    fig.update_layout(template='plotly_dark')
    fig.show()
    return fig


fig = plot_umap(plot_df, title='UMAP Projection: Game Semantic Similarity Map')
```

## Cluster the embedding

`cluster_points` runs DBSCAN over the 2D embedding to find density-based groups of similar games, and reports how many clusters were found and what fraction of points ended up as noise.

`assign_cluster_colors` turns those raw cluster ids into something a categorical palette can actually show: DBSCAN here finds hundreds of clusters, far more than the ~26 distinct colors in a qualitative palette, so cycling the palette made unrelated clusters look identical. Instead, only the `top_n` largest clusters get their own distinct color; every smaller cluster is folded into a single `"other"` bucket, and noise (`-1`) stays a faint gray, same as before.

- **embedding** — the 2D UMAP coordinates to cluster.
- **eps** — DBSCAN's neighborhood radius; smaller values find tighter, more numerous clusters.
- **min_samples** — minimum neighborhood size needed to seed a cluster.
- **cluster_labels** — raw per-point cluster id from DBSCAN (`-1` means noise).
- **n_clusters** — count of clusters DBSCAN found (excluding noise), returned alongside the labels for reuse in the plot title.
- **top_n** — how many of the largest clusters keep their own distinct color.
- **sizes** / **top_clusters** — cluster ids ranked by point count, used to pick which ones qualify for `top_n`.
- **cluster_ids** — string label per point after bucketing (a real cluster id, `"other"`, or `"-1"`), stored back into `plot_df['Cluster_ID']`.
- **color_discrete_map** — the `{cluster_id: color}` dict passed to `plot_umap` so colors stay stable and meaningful across reruns.

```python
def cluster_points(embedding, eps=0.3, min_samples=20):
    print("Computing Density-Based Clusters (DBSCAN)...")
    dbscan = DBSCAN(eps=eps, min_samples=min_samples, n_jobs=-1)
    cluster_labels = dbscan.fit_predict(embedding)

    n_clusters = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
    n_noise = list(cluster_labels).count(-1)
    print(f"Discovered Clusters: {n_clusters}")
    print(f"Noise Points: {n_noise} ({(n_noise / len(cluster_labels)) * 100:.1f}%)")

    return cluster_labels, n_clusters


def assign_cluster_colors(cluster_labels, top_n=25):
    sizes = pd.Series(cluster_labels).value_counts()
    sizes = sizes.drop(index=-1, errors='ignore')
    top_clusters = list(sizes.head(top_n).index)

    cluster_ids = pd.Series(cluster_labels).apply(
        lambda c: str(c) if (c == -1 or c in top_clusters) else 'other'
    )

    palette = px.colors.qualitative.Alphabet
    color_discrete_map = {
        str(c): palette[i % len(palette)] for i, c in enumerate(top_clusters)
    }
    color_discrete_map['other'] = 'rgba(120, 120, 120, 0.3)'
    color_discrete_map['-1'] = 'rgba(100, 100, 100, 0.1)'

    return cluster_ids, color_discrete_map


cluster_labels, n_clusters_found = cluster_points(umap_embedding)
plot_df['Cluster_ID'], color_discrete_map = assign_cluster_colors(cluster_labels)

fig_clusters = plot_umap(
    plot_df,
    color='Cluster_ID',
    color_discrete_map=color_discrete_map,
    title=f'DBSCAN Clusters (Noise Isolated) | {n_clusters_found} Clusters Found',
    opacity=1.0,
)
```
