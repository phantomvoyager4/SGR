```python
import pandas as pd
import numpy as np
import umap
import plotly.express as px
import matplotlib.pyplot as plt
import time
import seaborn as sns
from sklearn.cluster import MiniBatchKMeans
from sklearn.cluster import DBSCAN
from sklearn.decomposition import TruncatedSVD
import warnings
warnings.filterwarnings("ignore")
```

```python
def load_data(path='../../../data/parquet/games_fully_vectorized.parquet'):
    df = pd.read_parquet(path)
    return df

df = load_data()


potential_metadata = ['index', 'name']
metadata_cols = [col for col in potential_metadata if col in df.columns]

metadata_table = df[metadata_cols].copy()

vectors_table = df.drop(columns=metadata_cols)
```

```python
numerical_vectors = vectors_table.select_dtypes(include=['number'])

missing_cols = len(vectors_table.columns) - len(numerical_vectors.columns)
if missing_cols > 0:
    print(f"Removed {missing_cols} non-numeric/sequence columns from feature space.")

print(f"[{time.strftime('%X')}] Step 1/3: Reducing dimensions via TruncatedSVD...")
svd = TruncatedSVD(n_components=100, random_state=42)
vectors_reduced = svd.fit_transform(numerical_vectors)

print(f"[{time.strftime('%X')}] Step 2/3: Initializing UMAP with parallel processing...")
reducer = umap.UMAP(
    n_neighbors=15, 
    min_dist=0.1, 
    metric='cosine',
    init='spectral', 
    n_jobs=-1        
)

print(f"[{time.strftime('%X')}] Step 3/3: Fitting UMAP over {len(numerical_vectors):,} rows...")
umap_embedding = reducer.fit_transform(vectors_reduced)

plot_df = metadata_table.copy()
plot_df['UMAP_X'] = umap_embedding[:, 0]
plot_df['UMAP_Y'] = umap_embedding[:, 1]
print(f"[{time.strftime('%X')}] Processing complete. Ready for visualization.")
```

```python
fig = px.scatter(
    plot_df,
    x='UMAP_X',
    y='UMAP_Y',
    hover_name='name' if 'name' in plot_df.columns else None,
    hover_data=[col for col in metadata_cols if col != 'name'],
    title='UMAP Projection: Game Semantic Similarity Map',
    render_mode='webgl'
)

fig.update_traces(marker=dict(size=2, opacity=0.5))
fig.update_layout(template='plotly_dark')
fig.show()
```

```python
print("Computing Density-Based Clusters (DBSCAN)...")
dbscan = DBSCAN(eps=0.3, min_samples=20, n_jobs=-1)

cluster_labels = dbscan.fit_predict(umap_embedding)
plot_df['Cluster_ID'] = cluster_labels.astype(str)

n_clusters = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
n_noise = list(cluster_labels).count(-1)
print(f"Discovered Clusters: {n_clusters}")
print(f"Noise Points: {n_noise} ({(n_noise / len(cluster_labels)) * 100:.1f}%)")

color_map = {str(c): px.colors.qualitative.Alphabet[c % len(px.colors.qualitative.Alphabet)] for c in range(n_clusters)}
color_map['-1'] = 'rgba(100, 100, 100, 0.1)'

fig_clusters = px.scatter(
    plot_df,
    x='UMAP_X',
    y='UMAP_Y',
    color='Cluster_ID',
    hover_name='name' if 'name' in plot_df.columns else None,
    color_discrete_map=color_map,
    title=f'DBSCAN Clusters (Noise Isolated) | {n_clusters} Clusters Found',
    render_mode='webgl'
)

fig_clusters.update_traces(marker=dict(size=2))
fig_clusters.update_layout(template='plotly_dark')
fig_clusters.show()
```
